```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

n = int(input())
print(fibonacci(n))
```

```python
n = int(input())
nums = list(map(int, input().split()))

seen = set()
duplicate = False

for num in nums:
    if num in seen:
        duplicate = True
        break
    seen.add(num)

print("true" if duplicate else "false")
```

```python
m, n = map(int, input().split())

accounts = []

for _ in range(m):
    accounts.append(list(map(int, input().split())))

max_wealth = 0

for customer in accounts:
    wealth = sum(customer)
    if wealth > max_wealth:
        max_wealth = wealth

print(max_wealth)
```

```python
n = int(input())
nums = list(map(int, input().split()))

result = []

for i in range(n):
    count = 0
    for j in range(n):
        if i != j and nums[j] < nums[i]:
            count += 1
    result.append(count)

print(*result)
```

```python
s = input()

count = {}

for char in s:
    count[char] = count.get(char, 0) + 1

answer = -1

for i in range(len(s)):
    if count[s[i]] == 1:
        answer = i
        break

print(answer)
```
