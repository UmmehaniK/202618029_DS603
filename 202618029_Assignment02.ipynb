```python
#Problem 1: Largest, Smallest, Sum and Average of a List

ls=[]
for i in range(0,10):
    i=int(input("Add Number = "))
    ls.append(i)

print()
print("LIST = ",ls)
print("The Largest Element = ",max(ls))
print("The Smallest Element = ",min(ls))
print("The Sum of all Elements = ",sum(ls))
avg=(sum(ls)/len(ls))
print("The Average of the List = ",round(avg,2)) 
```

```python
#Problem 2: Days of the Week Using a Tuple

days = ("Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday")
print("First Three Days:", *days[:3])
mon, tue, wed, thu, fri, sat, sun = days

print()
print(mon)
print(tue)
print(wed)
print(thu)
print(fri)
print(sat)
print(sun)
```

```python
#Problem 3: Employee Details Using a Dictionary

employee_id = input()
name = input()
department = input()
initial_salary = int(input())
updated_salary = int(input())
email = input()

employee = {
    "EmployeeID": employee_id,
    "Name": name,
    "Department": department,
    "Salary": initial_salary
}

employee["Salary"] = updated_salary
employee["Email"] = email

for key, value in employee.items():
    print(key, ":", value)
```

```python
#Problem 4: Set Operations and Disjoint Check

n1 = int(input())
set1 = set(map(int, input().split()))

n2 = int(input())
set2 = set(map(int, input().split()))

union = set1 | set2
intersection = set1 & set2
difference = set1 - set2
symmetric_difference = set1 ^ set2

print("Union:", union)
print("Intersection:", intersection)
print("Difference:", difference)
print("Symmetric Difference:", symmetric_difference)
print("Disjoint:", "Yes" if set1.isdisjoint(set2) else "No")
```

```python
#Problem 5: Remove Duplicate Elements Using a Set

n = int(input())
numbers = list(map(int, input().split()))

seen = set()
unique_numbers = []

for number in numbers:
    if number not in seen:
        seen.add(number)
        unique_numbers.append(number)

print("Original List:", numbers)
print("List After Removing Duplicates:", unique_numbers)
```

```python
#Problem 6: Marks of Five Students Using a Dictionary

students = {}

for i in range(5):
    name, marks = input().split()
    students[name] = int(marks)

for name, marks in students.items():
    print(name, ":", marks)

average = sum(students.values()) / len(students)
highest_student = max(students, key=students.get)

print("Average Marks:", f"{average:.2f}")
print("Highest Marks:", highest_student, f"({students[highest_student]})")
```
