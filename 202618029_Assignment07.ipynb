# Problem 01 — Campus Student Registry

```python
class Student:
    university = "DS_University"
    student_count = 0

    def __init__(self, name, roll, cgpa):
        if not Student.is_valid_cgpa(cgpa):
            raise ValueError("Invalid CGPA")
        self.name = name
        self.roll = roll
        self.cgpa = cgpa
        Student.student_count += 1

    @classmethod
    def from_string(cls, text):
        name, roll, cgpa = text.split("-")
        return cls(name, roll, float(cgpa))

    @classmethod
    def from_dict(cls, data):
        return cls(data["name"], data["roll"], float(data["cgpa"]))

    @classmethod
    def change_university(cls, new_name):
        cls.university = new_name

    @staticmethod
    def is_valid_cgpa(cgpa):
        return 0.0 <= cgpa <= 10.0

    def letter_grade(self):
        if self.cgpa >= 9:
            return "A"
        if self.cgpa >= 8:
            return "B"
        if self.cgpa >= 7:
            return "C"
        if self.cgpa >= 6:
            return "D"
        return "F"

    def __str__(self):
        return f"{self.name} ({self.roll}) | CGPA: {self.cgpa:.2f} | Grade: {self.letter_grade()} | University: {self.university}"
```

```python
s1 = Student("Ravi", "101", 8.5)
s2 = Student.from_string("Meena-102-9.2")
s3 = Student.from_dict({"name": "Aman", "roll": "103", "cgpa": 7.4})

print(s1)
print(s2)
print(s3)
print("Count:", Student.student_count)
```

```python
Student.change_university("New_University")

s1.university = "Legacy_Institute"

print(s1)
print(s2)
print(s3)
print("Count:", Student.student_count)
```

```python
students = {}

n = int(input())

for _ in range(n):
    p = input().split()

    try:
        if p[0] == "ADD":
            s = Student(p[1], p[2], float(p[3]))
            students[p[2]] = s
            print(s)

        elif p[0] == "ADD_STR":
            s = Student.from_string(p[1])
            students[s.roll] = s
            print(s)

        elif p[0] == "ADD_DICT":
            s = Student.from_dict({
                "name": p[1],
                "roll": p[2],
                "cgpa": p[3]
            })
            students[s.roll] = s
            print(s)

        elif p[0] == "RENAME_UNI":
            Student.change_university(p[1])

        elif p[0] == "SHADOW":
            students[p[1]].university = p[2]

        elif p[0] == "INFO":
            print(students[p[1]])

        elif p[0] == "COUNT":
            print(Student.student_count)

        elif p[0] == "GRADE":
            print(students[p[1]].letter_grade())

    except ValueError:
        print("Error: Invalid CGPA")
```

# Problem 02 — Corporate Payroll Hierarchy.

```python
class Employee:
    role = "Employee"

    def __init__(self, name, base_salary):
        self.name = name
        self.base_salary = base_salary

    def monthly_pay(self):
        return self.base_salary

    def _extra(self):
        return ""

    def describe(self):
        return f"{self.name} [{self.role}{self._extra()}]"


class Manager(Employee):
    role = "Manager"

    def __init__(self, name, base_salary, team_size):
        super().__init__(name, base_salary)
        self.team_size = team_size

    def monthly_pay(self):
        return super().monthly_pay() + 500 * self.team_size

    def _extra(self):
        return f", team={self.team_size}"


class Director(Manager):
    role = "Director"

    def monthly_pay(self):
        return super().monthly_pay() * 1.20


class Engineer(Employee):
    role = "Engineer"

    def __init__(self, name, base_salary, level):
        super().__init__(name, base_salary)
        self.level = level

    def monthly_pay(self):
        return self.base_salary * (1 + 0.10 * self.level)

    def _extra(self):
        return f", level={self.level}"


class Intern(Employee):
    role = "Intern"

    def monthly_pay(self):
        return self.base_salary * 0.5
```

```python
n = int(input())
emp = {}

for _ in range(n):
    p = input().split()
    typ, name, sal = p[0], p[1], float(p[2])

    if typ == "Employee":
        e = Employee(name, sal)
    elif typ == "Manager":
        e = Manager(name, sal, int(p[3]))
    elif typ == "Director":
        e = Director(name, sal, int(p[3]))
    elif typ == "Engineer":
        e = Engineer(name, sal, int(p[3]))
    else:
        e = Intern(name, sal)

    emp[name] = e
    print(f"{e.describe()} | Pay: {e.monthly_pay():.2f}")

total = sum(e.monthly_pay() for e in emp.values())
high = max(emp.values(), key=lambda e: e.monthly_pay())

print(f"Total: {total:.2f}")
print(f"Highest: {high.name}")

q = int(input())

for _ in range(q):
    p = input().split()

    if p[0] == "ISA":
        print(isinstance(emp[p[1]], globals()[p[2]]))
    else:
        c = globals()[p[1]]
        print(" -> ".join(x.__name__ for x in c.__mro__))
```

# Problem 03 — Smart-Home Device Fabric

```python
class Device:
    feature = None

    def __init__(self, name):
        self.name = name
        self.log = []

    def power_on(self):
        self.log = ["Device core booted"]

    def mro_names(self):
        return " -> ".join(c.__name__ for c in type(self).__mro__)

    def capabilities(self):
        return [
            vars(c)["feature"]
            for c in type(self).__mro__
            if vars(c).get("feature") is not None
        ]


class Camera(Device):
    feature = "video"

    def power_on(self):
        super().power_on()
        self.log.append("Camera sensor online")


class Speaker(Device):
    feature = "audio"

    def power_on(self):
        super().power_on()
        self.log.append("Speaker amplifier online")


class Sensor(Device):
    feature = "motion"

    def power_on(self):
        super().power_on()
        self.log.append("Motion sensor armed")


class Doorbell(Camera, Speaker):
    feature = "chime"

    def power_on(self):
        super().power_on()
        self.log.append("Doorbell chime ready")


class SecurityHub(Doorbell, Sensor):
    feature = "uplink"

    def power_on(self):
        super().power_on()
        self.log.append("Hub uplink established")
```

```python
dev = {}

n = int(input())

for _ in range(n):
    p = input().split()
    typ, name = p[1], p[2]

    if typ == "Camera":
        d = Camera(name)
    elif typ == "Speaker":
        d = Speaker(name)
    elif typ == "Sensor":
        d = Sensor(name)
    elif typ == "Doorbell":
        d = Doorbell(name)
    elif typ == "SecurityHub":
        d = SecurityHub(name)
    else:
        d = Device(name)

    dev[name] = d

q = int(input())

for _ in range(q):
    p = input().split()

    if p[0] == "POWER_ON":
        dev[p[1]].power_on()
        print(dev[p[1]].log)

    elif p[0] == "MRO":
        print(dev[p[1]].mro_names())

    elif p[0] == "CAPS":
        print(", ".join(dev[p[1]].capabilities()))
```

# Problem 04 — Smart Shopping Cart

```python
class Item:
    def __init__(self, name, price, qty=1):
        self.name = name
        self.price = price
        self.qty = qty

    @property
    def subtotal(self):
        return self.price * self.qty

    def __eq__(self, other):
        return isinstance(other, Item) and self.name == other.name and self.price == other.price

    def __hash__(self):
        return hash((self.name, self.price))

    def __repr__(self):
        return f"Item('{self.name}', {self.price}, {self.qty})"


class Cart:
    def __init__(self, owner):
        self.owner = owner
        self.__items = []

    def add(self, *args):
        if len(args) == 1 and isinstance(args[0], Item):
            item = args[0]
        elif len(args) == 2:
            item = Item(args[0], args[1])
        elif len(args) == 3:
            item = Item(args[0], args[1], args[2])
        else:
            raise TypeError("Invalid arguments")

        for x in self.__items:
            if x.name == item.name:
                x.qty += item.qty
                return

        self.__items.append(item)

    def __len__(self):
        return len(self.__items)

    def __bool__(self):
        return bool(self.__items)

    def __contains__(self, name):
        return any(x.name == name for x in self.__items)

    def __iter__(self):
        return iter(self.__items)

    def __getitem__(self, key):
        if isinstance(key, int):
            return self.__items[key]

        for x in self.__items:
            if x.name == key:
                return x

        raise KeyError(key)

    def __call__(self, discount_percent=0):
        total = sum(x.subtotal for x in self.__items)
        return total * (1 - discount_percent / 100)

    def __repr__(self):
        return f"Cart(owner='{self.owner}', items={len(self.__items)})"

    def remove(self, name):
        for i, x in enumerate(self.__items):
            if x.name == name:
                del self.__items[i]
                return
        raise KeyError(name)
```

```python
cart = None

n = int(input())

for _ in range(n):
    p = input().split()

    if p[0] == "CART":
        cart = Cart(p[1])

    elif p[0] == "ADD":
        cart.add(p[1], float(p[2]), int(p[3]))

    elif p[0] == "ADD_ITEM":
        cart.add(Item(p[1], float(p[2]), int(p[3])))

    elif p[0] == "REMOVE":
        cart.remove(p[1])

    elif p[0] == "GET":
        try:
            print(cart[p[1]] if not p[1].isdigit() else cart[int(p[1])])
        except (KeyError, IndexError) as e:
            print(type(e).__name__)

    elif p[0] == "HAS":
        print(p[1] in cart)

    elif p[0] == "LEN":
        print(len(cart))

    elif p[0] == "EMPTY":
        print(not cart)

    elif p[0] == "LIST":
        print(list(cart))

    elif p[0] == "TOTAL":
        d = float(p[1]) if len(p) > 1 else 0
        print(f"{cart(d):.2f}")

    elif p[0] == "REPR":
        print(repr(cart))
```

# **Problem 05 — Mini Core-Banking Platform**

```python
from abc import ABC, abstractmethod


class Account(ABC):
    def __init__(self, aid, balance=0):
        self.id = aid
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance

    def _apply(self, delta):
        self.__balance += delta

    @abstractmethod
    def available_funds(self):
        pass

    @abstractmethod
    def monthly_update(self):
        pass


class SavingsAccount(Account):
    kind = "Savings"
    MONTHLY_RATE = 0.01

    def available_funds(self):
        return self.balance

    def monthly_update(self):
        interest = self.balance * self.MONTHLY_RATE
        self._apply(interest)
        return f"{self.id} (Savings): +${interest:.2f} interest -> ${self.balance:.2f}"


class CurrentAccount(Account):
    kind = "Current"
    OVERDRAFT_LIMIT = 500
    MONTHLY_FEE = 10

    def available_funds(self):
        return self.balance + self.OVERDRAFT_LIMIT

    def monthly_update(self):
        self._apply(-self.MONTHLY_FEE)
        return f"{self.id} (Current): -${self.MONTHLY_FEE:.2f} fee -> ${self.balance:.2f}"


class Customer:
    def __init__(self, cid, name):
        self.id = cid
        self.name = name
        self.accounts = {}

    def add_account(self, acc):
        self.accounts[acc.id] = acc


class Bank:
    bank_name = "DS Bank"
    total_transactions = 0

    def __init__(self):
        self.customers = {}
        self.accounts = {}

    @classmethod
    def rename(cls, name):
        cls.bank_name = name

    @staticmethod
    def valid_amount(x):
        return x > 0

    def open(self, cid, name, typ, aid, balance):
        if cid not in self.customers:
            self.customers[cid] = Customer(cid, name)

        cls = {
            "SAVINGS": SavingsAccount,
            "CURRENT": CurrentAccount
        }[typ]

        acc = cls(aid, balance)
        self.customers[cid].add_account(acc)
        self.accounts[aid] = acc

    def deposit(self, aid, amount):
        if aid not in self.accounts:
            print("Account not found")
            return

        if not self.valid_amount(amount):
            print("Invalid amount")
            return

        self.accounts[aid]._apply(amount)
        Bank.total_transactions += 1

    def withdraw(self, aid, amount):
        if aid not in self.accounts:
            print("Account not found")
            return

        if not self.valid_amount(amount):
            print("Invalid amount")
            return

        acc = self.accounts[aid]

        if amount > acc.available_funds():
            print("Insufficient funds")
            return

        acc._apply(-amount)
        Bank.total_transactions += 1

    def transfer(self, a, b, amount):
        if a not in self.accounts or b not in self.accounts:
            print("Account not found")
            return

        if not self.valid_amount(amount):
            print("Invalid amount")
            return

        if amount > self.accounts[a].available_funds():
            print("Insufficient funds")
            return

        self.accounts[a]._apply(-amount)
        self.accounts[b]._apply(amount)
        Bank.total_transactions += 1
```

```python
bank = Bank()

n = int(input())

for _ in range(n):
    p = input().split()

    if p[0] == "OPEN":
        bank.open(p[1], p[2], p[3], p[4], float(p[5]))

    elif p[0] == "DEPOSIT":
        bank.deposit(p[1], float(p[2]))

    elif p[0] == "WITHDRAW":
        bank.withdraw(p[1], float(p[2]))

    elif p[0] == "TRANSFER":
        bank.transfer(p[1], p[2], float(p[3]))

    elif p[0] == "MONTH_END":
        for acc in bank.accounts.values():
            print(acc.monthly_update())

    elif p[0] == "RENAME":
        Bank.rename(p[1])

    elif p[0] == "SUMMARY":
        print(Bank.bank_name)
        for cid, c in bank.customers.items():
            for acc in c.accounts.values():
                print(f"{cid} {c.name} {acc.id} ({acc.kind}) ${acc.balance:.2f}")

    elif p[0] == "STATS":
        print(Bank.total_transactions)
```
